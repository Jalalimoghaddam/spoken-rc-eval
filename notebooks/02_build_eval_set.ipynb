```python
import re
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Raw messy data
raw_sentences = [
    "  ADMINISTER 500MG Orally Twice Daily!!  ",
    "administer 500 mg orally twice daily",
    "Patient has NO chest pain.",
    "Patient has chest pain.",
]

# 1) Cleaning
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

cleaned = [clean_text(s) for s in raw_sentences]

# 2) Tokenization
tokenized = [s.split() for s in cleaned]

# 3) Stopword removal — naive vs domain-aware
naive_stopwords = {"a", "the", "is", "has", "no", "not", "of", "to"}
medical_safe_stopwords = naive_stopwords - {"no", "not"}  # keep negation!

def remove_stopwords(tokens, stopword_set):
    return [t for t in tokens if t not in stopword_set]

for t in tokenized:
    print("naive:       ", remove_stopwords(t, naive_stopwords))
    print("domain-aware:", remove_stopwords(t, medical_safe_stopwords))
```

```python
paraphrase_pairs = [
    ("It was founded in 1852", "The museum opened in 1852"),
    ("45–60 nanometers across", "They measure between 45 and 60 nanometers in size"),
    ("considerable impact", "a significant effect"),
    ("gelatinous projections edged with cilia that produce water currents", "jelly-like structures lined with cilia that generate currents in the water"),
    ("lack of understanding of the legal ramifications", "not knowing the legal consequences"),
]

for original, paraphrase in paraphrase_pairs:
    emb1 = model.encode(original)
    emb2 = model.encode(paraphrase)
    sim = util.cos_sim(emb1, emb2).item()
    print(f"{sim:.3f}  |  {original}  <->  {paraphrase}")
```

```python
wrong_pairs = [
    ("It was founded in 1852", "Levi's Stadium"),
    ("45–60 nanometers across", "Jesus the Interpreter"),
    ("considerable impact", "Thoreau"),
    ("gelatinous projections edged with cilia that produce water currents", "50"),
    ("lack of understanding of the legal ramifications", "Mi'kmaq and the Abenaki"),
]

for a, b in wrong_pairs:
    emb1 = model.encode(a)
    emb2 = model.encode(b)
    sim = util.cos_sim(emb1, emb2).item()
    print(f"{sim:.3f}  |  {a}  <->  {b}")
```

```python
near_miss_pairs = [
    ("It was founded in 1852", "It was founded in 1920"),  # سال اشتباه
    ("45–60 nanometers across", "20–30 nanometers across"),  # عدد اشتباه
    ("considerable impact", "minimal impact"),  # صفت مخالف
    ("Levi's Stadium", "Soldier Field"),  # اسم استادیوم اشتباه
    ("Mi'kmaq and the Abenaki", "Cherokee and the Sioux"),  # اسم قبیله اشتباه
]

for original, near_miss in near_miss_pairs:
    emb1 = model.encode(original)
    emb2 = model.encode(near_miss)
    sim = util.cos_sim(emb1, emb2).item()
    print(f"{sim:.3f}  |  {original}  <->  {near_miss}")
```

```python
results = []

for original, paraphrase in paraphrase_pairs:
    sim = util.cos_sim(model.encode(original), model.encode(paraphrase)).item()
    results.append({"type": "paraphrase", "text1": original, "text2": paraphrase, "similarity": sim})

for a, b in wrong_pairs:
    sim = util.cos_sim(model.encode(a), model.encode(b)).item()
    results.append({"type": "wrong", "text1": a, "text2": b, "similarity": sim})

for a, b in near_miss_pairs:
    sim = util.cos_sim(model.encode(a), model.encode(b)).item()
    results.append({"type": "near_miss", "text1": a, "text2": b, "similarity": sim})

results_df = pd.DataFrame(results)
results_df.to_csv("../data/similarity_experiment.csv", index=False)
print(results_df)
```

```python
import re

def extract_numbers(text):
    # فقط ویرگول‌هایی که دقیقاً وسط یه عدد بزرگ هستن رو حذف کن
    # مثال: "1,893" -> "1893"   ولی   "1,2,3" باقی می‌مونه دست‌نخورده
    text = re.sub(r'(\d),(\d{3})', r'\1\2', text)
    numbers = re.findall(r'\d+', text)
    return set(numbers)
```

```python
print(extract_numbers("1,893"))
print(extract_numbers("1,2,3"))
print(extract_numbers("10,000"))
```

```python
ref_numbers = extract_numbers("45–60 nanometers across")
trans_numbers = extract_numbers("20–30 nanometers across")

print(ref_numbers)
print(trans_numbers)
print(ref_numbers == trans_numbers)
```

```python
import re
from sentence_transformers import util

def extract_numbers(text):
    text = re.sub(r'(\d),(\d{3})', r'\1\2', text)
    numbers = re.findall(r'\d+', text)
    return set(numbers)

def hybrid_similarity(reference_answer, transcript, model):
    semantic_sim = util.cos_sim(model.encode(reference_answer), model.encode(transcript)).item()
    ref_numbers = extract_numbers(reference_answer)
    trans_numbers = extract_numbers(transcript)

    print("DEBUG ref_numbers:", ref_numbers)
    print("DEBUG trans_numbers:", trans_numbers)
    print("DEBUG equal?:", ref_numbers == trans_numbers)

    if len(ref_numbers) == 0:
        return semantic_sim
    else:
        if ref_numbers == trans_numbers:
            return semantic_sim
        else:
            return 0

result = hybrid_similarity("45–60 nanometers across", "20–30 nanometers across", model)
print("FINAL RESULT:", result)
```

```python
def format_timestamp(seconds):
    hours = int(seconds // 3600)
    remainder = seconds % 3600
    minutes = int(remainder // 60)
    secs = remainder % 60
    whole_secs = int(secs)
    milliseconds = int((secs - whole_secs) * 1000)
    return f"{hours:02d}:{minutes:02d}:{whole_secs:02d},{milliseconds:03d}"
    
```

```python
print(format_timestamp(13.76))
print(format_timestamp(3725.5))
```

```python
segments, info = model.transcribe("data/audio/A2_travelling_abroad.mp3")

with open("data/audio/A2_travelling_abroad.srt", "w", encoding="utf-8") as f:
    for i, segment in enumerate(segments, start=1):
        start = format_timestamp(segment.start)
        end = format_timestamp(segment.end)
        f.write(f"{i}\n{start} --> {end}\n{segment.text.strip()}\n\n")
```

```python
import pandas as pd

df = pd.read_csv("data/eval_set.csv")
df_shuffled = df.sample(frac=1, random_state=42).reset_index(drop=True)

gtts_batch = df_shuffled.iloc[:25]
pyttsx3_batch = df_shuffled.iloc[25:]

print(gtts_batch.shape)
print(pyttsx3_batch.shape)
```

```python
from gtts import gTTS

tts = gTTS(text="Saint Bernadette Soubirous", lang="en")
tts.save("test.mp3")
```

```python
import os
from gtts import gTTS

os.makedirs("data/audio/gtts", exist_ok=True)

for i, row in gtts_batch.iterrows():

    # 1. create a gTTS object using row["answer"] as the text
    tts = gTTS(text=row["answer"], lang="en")

    # 2. save it to a file like "data/audio/gtts/gtts_{i}.mp3"
    tts.save(f"data/audio/gtts/gtts_{i}.mp3")
```

```python
print(os.listdir("data/audio/gtts"))
```

```python
import pyttsx3

engine = pyttsx3.init()
os.makedirs("data/audio/pyttsx3", exist_ok=True)

for i, row in pyttsx3_batch.iterrows():
    engine.save_to_file(row["answer"], f"data/audio/pyttsx3/pyttsx3_{i}.mp3")

engine.runAndWait()
```

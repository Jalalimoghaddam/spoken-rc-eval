```python
def format_timestamp(seconds):
    hours = int(seconds // 3600)
    remainder = seconds % 3600
    minutes = int(remainder // 60)
    secs = remainder % 60
    whole_secs = int(secs)
    milliseconds = int((secs - whole_secs) * 1000)
    return f"{hours:02d}:{minutes:02d}:{whole_secs:02d},{milliseconds:03d}"
    
```

```python
print(format_timestamp(13.76))
print(format_timestamp(3725.5))
```

```python
segments, info = model.transcribe("data/audio/A2_travelling_abroad.mp3")

with open("data/audio/A2_travelling_abroad.srt", "w", encoding="utf-8") as f:
    for i, segment in enumerate(segments, start=1):
        start = format_timestamp(segment.start)
        end = format_timestamp(segment.end)
        f.write(f"{i}\n{start} --> {end}\n{segment.text.strip()}\n\n")
```
